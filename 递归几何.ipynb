```python
def F(n):
    if(n==0):
        return 1
    else :
        return n-M(F(n-1))
def M(n):
    if(n==0):
        return 0
    else:
        return n-F(M(n-1))
def main():
    print("F")
    for i in range(10 ):
       print(F(i))
    print("M")
    for i in range(15):
       print(M(i))
```

```python
main()
```
